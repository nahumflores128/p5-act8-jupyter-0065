```python
datos = {
    "Producto": [" Arroz", "Pizza", "Hamburguesa", "Chilaquiles", "Tacos"],
    "Precio": [35, 20, 50, 40, 15],
    "Cantidad": [1, 9, 1, 9, 3]
}

df = pd.DataFrame(datos)

df["Total"] = df["Precio"] * df["Cantidad"]

print("=== VENTAS DE COMIDA ===")
print(df)
print("+-+-++-+-jupyter Nahum++-+-+-++--")
print("Nahum flores NC 0065")
```

```python
print("Productos con menos de 10 unidades:")
print(df[df["Existencias"] < 10])

print("\nExistencias totales:", df["Existencias"].sum())
print("Nahum flores NC 0065")

```
